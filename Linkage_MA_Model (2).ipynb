# Forward/Inverse Position Kinematics for a 2-member 2-DOF arm

```python
# simple forward kinematics for a 2 DOF arm

def solve_limb(lengths, angles):
    pts = np.zeros((len(lengths), 2))
    pos = np.zeros(2)
    ang = 0.0

    for i in range(len(lengths)):
        ang += angles[i]
        pos = pos + np.array([
            lengths[i] * np.cos(ang),
            lengths[i] * np.sin(ang)
        ])
        pts[i] = pos

    return pts

def solve_toe(lengths, angles):
    return solve_limb(lengths, angles)[-1]
```

```python
# inverse kinematics for a 2 DOF arm
def jacobian(a, b, h, k):
  return np.array([
      [-a*np.sin(h) - b*np.sin(h+k), -b*np.sin(h+k)],
      [ a*np.cos(h) + b*np.cos(h+k),  b*np.cos(h+k)]
  ])

def position_solve_jacobian(a, b, task_local, q0, step=0.5, timeout=1000, thresh=1e-4):
    q = np.array(q0, dtype=float)
    lengths = np.array([a, b], dtype=float)

    for _ in range(timeout):
      toe = solve_toe(lengths, q)
      err = task_local - toe

      if np.linalg.norm(err) < thresh:
          break

      J = jacobian(a, b, q[0], q[1])
      dq = np.linalg.pinv(J) @ err
      q = q + step * dq

    return q

def velocity_solve_jacobian(a, b, xdot_des, q, qdot0):
  q = np.array(q, dtype=float)
  qdot = np.array(qdot0, dtype=float)

  J = jacobian(a, b, q[0], q[1])
  J_inv = np.linalg.pinv(J) @ xdot_des
  qdot = J_inv @ xdot_des

  return qdot

def acceleration_solve_jacobian(a, b, xdotdot_des, q, qdotdot0):
  q = np.array(q, dtype=float)
  qdotdot = np.array(qdotdot0, dtype=float)

  J = jacobian(a, b, q[0], q[1])
  J_inv = np.linalg.pinv(J) @ xdotdot_des
  qdotdot = J_inv @ xdotdot_des

  return qdotdot

def solve_arm_position_world(a, b, base, task_world, q0):
  task_local = np.array(task_world) - np.array(base)
  return position_solve_jacobian(a, b, task_local, q0)

def solve_arm_velocity_world(a, b, base, xdot_des_world, q0, qdot0):
  xdot_des_local = np.array(xdot_des_world) - np.array(base)
  return velocity_solve_jacobian(a, b, xdot_des_local, qdot0)

# Source - https://stackoverflow.com/a/55817881
# Posted by mujjiga, modified by community. See post 'Timeline' for change history
# Retrieved 2026-03-26, License - CC BY-SA 4.0
def circle_circle_intersections(c1, c2, r):
    # circle 1: (x0, y0), radius r0
    # circle 2: (x1, y1), radius r1

    x0, y0 = c1
    x1, y1 = c2
    d=np.sqrt((x1-x0)**2 + (y1-y0)**2)

    # non intersecting
    if d > 2 * r:
        return None
    # One circle within other
    if d < 0:
        return None
    # coincident circles
    if d == 0:
        return None
    else:
        a=(d**2)/(2*d)
        h=np.sqrt(r**2-a**2)
        x2=x0+a*(x1-x0)/d
        y2=y0+a*(y1-y0)/d
        x3=x2+h*(y1-y0)/d
        y3=y2-h*(x1-x0)/d

        x4=x2-h*(y1-y0)/d
        y4=y2+h*(x1-x0)/d

        P1 = np.array((x3, y3))
        P2 = np.array((x4, y4))

        return P1, P2
```

```python
# methods for solution visualization
def plot_limb(lengths, angles, base=(0,0), plot_style='b-'):
    prev = np.array(base, dtype=float)
    ang = 0.0

    for i, L in enumerate(lengths):
        ang += angles[i]
        joint = prev + np.array([L*np.cos(ang), L*np.sin(ang)])
        plt.plot([prev[0], joint[0]], [prev[1], joint[1]], plot_style)
        prev = joint

def plot_task(task):
  x, y = task
  plt.plot(x, y, "ro")
```

# Inverse Position Kinematics for 5-Bar Linkage
Model: solve vector lengths for two linear actuators that constrain each other. Then, solve 2-member IK for those linear actuators.

```python
def get_S(P, C):
  # this is just simple vector math!
  # P is input, C is a system parameter
  return P, P - C
```

```python
def get_toggled_arm_solution_branch(q, L, desired_sign):
  # Take cross product of (elbow - base) and (end - elbow)
  # Assume local, so base is origin.
  limb = solve_limb(L, q)
  E = limb[0]
  P = limb[1]
  EP = P - E

  z = np.cross(EP, E)
  z_sign = np.sign(z)

  if z_sign != desired_sign:
    # change to the other config using geometry
    q2 = -q[1]

    A = L[0] + L[1] * np.cos(q2)
    B = L[1] * np.sin(q2)

    x, y = P
    q1 = np.atan2(y, x) - np.atan2(B, A)

    return np.array([q1, q2])
  else:
    return q
```

```python
# example
c_length = 0.1 # m
b_length = 0.1 # m
f_length = 0.2 # m
L = np.array([b_length, f_length])

# arm bases
B1 = np.array([0.0, 0.0])
B2 = np.array([c_length, 0.0])

# task position
P = np.array((0.1, 0.2))

# Solve 2-member IK as tasks
q1 = solve_arm_position_world(b_length, f_length, B1, P, q0=[0.3, 0.3])
q2 = solve_arm_position_world(b_length, f_length, B2, P, q0=[2.8, -0.3])

# Ensure arms are on the correct branches using cross products.
# One arm should be "up" and one should be "down"
q1 = get_toggled_arm_solution_branch(q1, L, 1)
q2 = get_toggled_arm_solution_branch(q2, L, -1)

# to demonstrate the alternative, show the incorrect version
q1_prime = get_toggled_arm_solution_branch(q1, L, -1)
q2_prime = get_toggled_arm_solution_branch(q2, L, 1)

print(f'Left angles:\tθ={round(q1[0], 4)}\tφ={round(q1[1], 4)}')
print(f'Right angles:\tθ={round(q2[0], 4)}\tφ={round(q2[1], 4)}')
```

# Forward/Inverse Kinematics for 5-bar linkage (Actuated Velocity & Acceleration Model)
This is a closed-loop solution for finding actuator output velocity/acceleration as a function of input ONLY.

This is a simplified version of an approach I developed at Sung Lab last summer to cleanly mechanical integrate constraints into inverse kinematics on systems with partial actuation.

```python
# define constraints and generalized constraint Jacobian (Jc) rows
def link_constraint(joint_A, joint_B, length, positions):
  r0 = positions[joint_A]
  r1 = positions[joint_B]
  return length - np.linalg.norm(r0 - r1)

def fixed_x(joint, fix_x, positions):
  r0 = positions[joint]
  return fix_x - r0[0]

def fixed_y(joint, fix_y, positions):
  r0 = positions[joint]
  return fix_y - r0[1]

def del_link_constraint(joint_A, joint_B, length, positions, partials):
  r0 = positions[joint_A]
  r1 = positions[joint_B]
  del_r0 = partials[joint_A]
  del_r1 = partials[joint_B]

  diff = r1 - r0
  norm_diff = np.linalg.norm(diff)

  return (diff / norm_diff) @ (del_r1 - del_r0)

def del_fixed_x(joint, positions, partials):
  del_r0 = partials[joint]
  return -del_r0[0]

def del_fixed_y(joint, positions, partials):
  del_r0 = partials[joint]
  return -del_r0[1]

def get_r(joint, positional_q):
    idx = joint
    return np.array(positional_q[2*idx:2*(idx+1)])

def get_del_r(idx, positional_q):
  row = np.zeros((2, positional_q.shape[0]))
  row[:, 2*idx:2*(idx+1)] = np.eye(2)
  return row

def get_positions_and_partials(positional_q, n=5):
  positions = np.array([get_r(joint, positional_q) for joint in range(n)])
  partials = np.array([get_del_r(joint, positional_q) for joint in range(n)])
  return positions, partials
```

```python
# Model the linkage as a pentagon. Each vertex has an index.
# The origin vertex is 0, and indices increment clockwise

def get_positional_q(q1_full, q2_full, B1, B2, L):
  # turn q1_full and q2_full into the position q
  limb1_local = solve_limb(L, q1_full)
  limb2_local = solve_limb(L, q2_full)

  elbow_1, end_1 = limb1_local + B1
  elbow_2, _ = limb2_local + B2

  # flatten these limbs
  q = np.concatenate((B1, elbow_1, end_1, elbow_2, B2))
  return q.flatten()

def build_position_residuals(positions, L, n=5):
  # link constraints (constrains 5 DOF)
  residuals = []
  for i, length in enumerate(L):
    residuals.append(link_constraint(i%n, (i+1)%n, length, positions))

  # motor locations are fixed, so constrain their x and y (constrains 3 DOF)
  residuals.append(fixed_x(0, positions[0][0], positions))
  residuals.append(fixed_y(0, positions[0][1], positions))
  residuals.append(fixed_y(4, positions[4][1], positions))
  residuals.append(fixed_y(4, positions[4][0], positions)) #<-- redundant with length constraint!

  return np.array(residuals)

def build_geometric_jacobian(positions, partials, L, n=5):
  # This maps del residuals / del q. Each col is a different q_i
  # add link constraints (constrains 5 DOF)
  jacobian = []
  for i, length in enumerate(L):
    jacobian.append(del_link_constraint(i % n, (i + 1) % n, length, positions, partials))

  # motor locations are fixed, so constrain their x and y (constrains 3 DOF)
  jacobian.append(del_fixed_x(0, positions, partials))
  jacobian.append(del_fixed_y(0, positions, partials))
  jacobian.append(del_fixed_y(4, positions, partials))
  jacobian.append(del_fixed_x(4, positions, partials)) #<-- redundant with length constraint!

  return np.vstack(jacobian)

def solve_5_bar_velocities(theta_1_dot, theta_2_dot, positional_q, Ls_path):
  # Given motor speeds, find joint velocities
  positions, pos_partials = get_positions_and_partials(positional_q)
  J_complete = build_geometric_jacobian(positions, pos_partials, Ls_path)

  # coordinate indices, not joint indices
  elbow_coord_idxs = np.array([2, 3, 6, 7])   # joints 1 and 3 in 2D
  rest_coord_idxs  = np.array([0, 1, 4, 5, 8, 9])

  J_geom_rest  = J_complete[:, rest_coord_idxs]
  J_geom_elbow = J_complete[:, elbow_coord_idxs]

  theta_1 = np.atan2(positions[1][1] - positions[0][1],
                      positions[1][0] - positions[0][0])
  theta_2 = np.atan2(positions[3][1] - positions[4][1],
                      positions[3][0] - positions[4][0])

  b = Ls_path[-1]   # assuming proximal length

  # elbow coordinate velocities from motor speeds
  q_dot_elbow = np.array([
      -b * np.sin(theta_1) * theta_1_dot,
        b * np.cos(theta_1) * theta_1_dot,
      -b * np.sin(theta_2) * theta_2_dot,
        b * np.cos(theta_2) * theta_2_dot
  ])

  q_dot_rest = -np.linalg.pinv(J_geom_rest) @ (J_geom_elbow @ q_dot_elbow)
  q_dot_end_effector = q_dot_rest[2:4] # middle 2 points of 6
  q_dot = np.concatenate((q_dot_rest[0:2], q_dot_elbow[0:2], q_dot_end_effector, q_dot_elbow[2:4], q_dot_rest[4:6]))
  return q_dot_end_effector, q_dot, J_geom_rest, J_geom_elbow

def solve_5_bar_angular_speeds(task_xdot, task_ydot, positional_q, Ls_path):
  # Given velocities at P, find motor speeds
  positions, pos_partials = get_positions_and_partials(positional_q)
  J_complete = build_geometric_jacobian(positions, pos_partials, Ls_path)

  # determine current angles
  theta_1 = np.atan2(positions[1][1] - positions[0][1],
                      positions[1][0] - positions[0][0])
  theta_2 = np.atan2(positions[3][1] - positions[4][1],
                      positions[3][0] - positions[4][0])

  b = Ls_path[-1]

  # coordinate indices, not joint indices
  end_coord_idxs = np.array([4, 5])   # joints 3 in 2D
  rest_coord_idxs  = np.array([0, 1, 2, 3, 6, 7, 8, 9])
  elbow_coord_idxs = np.array([2, 3, 6, 7])   # joints 1 and 3 in 2D

  J_geom_rest  = J_complete[:, rest_coord_idxs]
  J_geom_end = J_complete[:, end_coord_idxs]

  # determine full qdot from task
  q_dot_end = np.array([task_xdot, task_ydot]) # task
  q_dot_rest = -np.linalg.pinv(J_geom_rest) @ (J_geom_end @ q_dot_end)
  q_dot = np.concatenate((q_dot_rest[0:4], q_dot_end, q_dot_rest[4:8]))

  # extract elbow from q_dot
  q_dot_elbow = q_dot[elbow_coord_idxs] # velocities, we want angular
  v1 = q_dot_elbow[0:2]
  v2 = q_dot_elbow[2:4]

  t1 = np.array([-np.sin(theta_1), np.cos(theta_1)])
  t2 = np.array([-np.sin(theta_2), np.cos(theta_2)])

  theta_1_dot = np.dot(v1, t1) / b
  theta_2_dot = np.dot(v2, t2) / b
  q_ang = np.array([theta_1_dot, theta_2_dot]) # this is the solution

  return q_ang, q_dot, J_geom_rest, J_geom_end
```

# 5-bar Position-Speed-Force Profile
How do rotational speeds and torques at the inputs relate to linear speeds and forces at the output?

From:
x_dot = J(q) * q_dot, and
x_dotdot = J(q) * q_dotdot

|J(q)| can be thought of as how q_dot or q_dotdot scales into positional movement.

However, since only θ is actuated, we actually don't care about the magnitudes of the φ rows. We thus ignore them.


```python
# determine workspace bounds
table_width_ft = 3 # ft
table_width = table_width_ft / 3.281 # m
workspace_depth_offset = 0.3 # m
workspace_depth_ft = 1.5 # ft
workspace_depth = workspace_depth_ft / 3.281 + workspace_depth_offset # m

# create sample points within the workspace
N = 5
X = np.linspace(0, table_width, N)
Y = np.linspace(workspace_depth_offset, workspace_depth, N)
xv, yv = np.meshgrid(X, Y)

# set up the arm
c_length = 0.1 # m
b_length = 0.2 # m
f_length = 0.3 # m
L = np.array([b_length, f_length])
center = table_width / 2
B1 = np.array([center - c_length / 2, 0.0])
B2 = np.array([center + c_length / 2, 0.0])

# input motor speeds
theta_1_dot = 1
theta_2_dot = 1
```

#Linkage Length Model

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.colors import LinearSegmentedColormap
from itertools import product as iproduct

FIELD_W    = 24.0   # field width (x)
FIELD_H    = 48.0   # full field depth (y)  — 2 × 4 ft
TARGET_Y0  = 0.0   # y-start of target zone
TARGET_Y1  = 24.0   # y-end   of target zone
TARGET_W   = 24.0
TARGET_H   = TARGET_Y1 - TARGET_Y0  # 24 in

MOTOR_Y    = -3    # motors sit 2 in from back wall
MOTOR_SEP  = 10    # separation between the two motors
MOTOR_LX   = (FIELD_W - MOTOR_SEP) / 2
MOTOR_RX   = (FIELD_W + MOTOR_SEP) / 2
MOTOR_L    = np.array([MOTOR_LX, MOTOR_Y])
MOTOR_R    = np.array([MOTOR_RX, MOTOR_Y])

THETA_STEPS = 150
GRID_RES   = 0.25   # inches per cell (0.25 in → 96×96 cells over 24×24 target)
EPS_DIST   = 1e-6   # minimum elbow separation before "full-fold" singularity
EPS_ELBOW  = 1e-4   # cos α proximity to ±1 before "elbow-lock" singularity
EPS_JAC    = 0.05   # |det C|/L2 normalized threshold for output singularity

def fk_5bar(theta_L, theta_R, L1, L2):
    """
    Compute end-effector position for given joint angles and link lengths,
    with full singularity checking.

    Singularity types checked:
      - Full-fold      : elbows A, B nearly coincident  (d ≈ 0)
      - Full-extension : elbow triangle completely flat  (d ≈ 2·L2)
      - Elbow-lock     : cos α very close to ±1          (degenerate triangle)
      - Output sing.   : end-effector constraint matrix C is rank-deficient
                         (robot instantaneously loses a DOF)

    Returns
    -------
    E           : np.ndarray (2,) end-effector position, or None if infeasible
    is_singular : bool  — True if near any singularity
    reason      : str   — description of singularity ('' if none)
    """
    A = MOTOR_L + L1 * np.array([np.cos(theta_L), np.sin(theta_L)])
    B = MOTOR_R + L1 * np.array([np.cos(theta_R), np.sin(theta_R)])

    d = np.linalg.norm(B - A)

    if d < EPS_DIST:
        return None, True, "full-fold: elbows coincident"

    if d > 2.0 * L2:
        return None, False, "unreachable: d > 2·L2"

    cos_alpha = d / (2.0 * L2)
    if abs(cos_alpha) >= 1.0 - EPS_ELBOW:
        return None, True, "full-extension or elbow-lock: degenerate triangle"

    alpha = np.arccos(np.clip(cos_alpha, -1.0, 1.0))

    AB   = (B - A) / d
    perp = np.array([-AB[1], AB[0]])
    mid  = (A + B) / 2.0
    h    = L2 * np.sin(alpha)

    E1 = mid + h * perp
    E2 = mid - h * perp
    E = E1 if E1[1] >= E2[1] else E2
    EA = E - A
    EB = E - B
    det_C = EA[0] * EB[1] - EA[1] * EB[0]
    if abs(det_C) / (L2 * L2) < EPS_JAC:
        return E, True, f"output singularity: A-E-B collinear (|det C|={abs(det_C):.2e})"

    return E, False, ""

def compute_workspace(L1, L2, theta_steps=THETA_STEPS):
    """
    Sweep all (θ_L, θ_R) combinations and classify each reachable point.

    Returns
    -------
    valid_pts    : (N, 2) non-singular reachable positions
    singular_pts : (M, 2) positions that lie on or near a singularity
    singular_log : list of (x, y, reason) for each singular config
    """
    angles       = np.linspace(0, 2 * np.pi, theta_steps, endpoint=False)
    valid_pts    = []
    singular_pts = []
    singular_log = []

    for tL, tR in iproduct(angles, angles):
        E, is_sing, reason = fk_5bar(tL, tR, L1, L2)
        if E is None:
            continue
        if is_sing:
            singular_pts.append(E)
            singular_log.append((E[0], E[1], reason))
        else:
            valid_pts.append(E)

    vp = np.array(valid_pts)    if valid_pts    else np.empty((0, 2))
    sp = np.array(singular_pts) if singular_pts else np.empty((0, 2))
    return vp, sp, singular_log


def coverage_fraction(valid_points, res=GRID_RES):
    """
    Fraction of the 24×24 in target zone covered by valid (non-singular) points.
    """
    if len(valid_points) == 0:
        return 0.0

    nx = int(np.ceil(TARGET_W / res))
    ny = int(np.ceil(TARGET_H / res))
    grid = np.zeros((nx, ny), dtype=bool)

    for x, y in valid_points:
        lx = x
        ly = y - TARGET_Y0
        ix = int(lx / res)
        iy = int(ly / res)
        if 0 <= ix < nx and 0 <= iy < ny:
            grid[ix, iy] = True

    return grid.sum() / (nx * ny)

def search_linkage_lengths(
    L1_range=(4, 26),
    L2_range=(4, 26),
    n_samples=15,
    theta_steps=120,
    verbose=True,
):
    """
    Grid-search (L1, L2) to maximise coverage of the 24×24 target zone.

    Returns
    -------
    best    : dict  {L1, L2, coverage, valid_pts, singular_pts, singular_log}
    results : list  of (L1, L2, coverage) for every (L1, L2) pair tried
    """
    L1_vals = np.linspace(*L1_range, n_samples)
    L2_vals = np.linspace(*L2_range, n_samples)
    total   = n_samples * n_samples

    best = {
        "L1": None, "L2": None, "coverage": -1.0,
        "valid_pts": None, "singular_pts": None, "singular_log": [],
    }
    results = []

    for i, (L1, L2) in enumerate(iproduct(L1_vals, L2_vals)):
        vp, sp, slog = compute_workspace(L1, L2, theta_steps=theta_steps)
        cov = coverage_fraction(vp)
        results.append((L1, L2, cov))

        if verbose and (i % max(1, total // 20) == 0):
            n_total = len(vp) + len(sp)
            sing_pct = len(sp) / n_total if n_total else 0
            print(f"  [{i+1:4d}/{total}]  L1={L1:5.2f}\"  L2={L2:5.2f}\"  "
                  f"coverage={cov:.1%}  singular={sing_pct:.1%}")

        if cov > best["coverage"]:
            best = {
                "L1": L1, "L2": L2, "coverage": cov,
                "valid_pts": vp, "singular_pts": sp, "singular_log": slog,
            }

    return best, results


def fine_tune(best, margin=2.0, n_samples=20, theta_steps=150, verbose=True):
    """Narrow search in a window around the coarse best solution."""
    print("\n── Fine-tuning ──")
    L1_range = (max(1.0, best["L1"] - margin), best["L1"] + margin)
    L2_range = (max(1.0, best["L2"] - margin), best["L2"] + margin)
    return search_linkage_lengths(L1_range, L2_range, n_samples, theta_steps, verbose)
def print_report(best):
    L1, L2 = best["L1"], best["L2"]
    cov    = best["coverage"]
    vp     = best["valid_pts"]
    sp     = best["singular_pts"]
    slog   = best["singular_log"]

    print("\n" + "═" * 62)
    print("  5-BAR LINKAGE OPTIMAL SOLUTION REPORT")
    print("═" * 62)
    print(f"  Full field size     : {FIELD_W:.0f} × {FIELD_H:.0f} in  (2 × 4 ft)")
    print(f"  Target zone (front) : {FIELD_W:.0f} × {TARGET_H:.0f} in  (2 × 2 ft)")
    print(f"  Motor L             : ({MOTOR_LX:.1f}, {MOTOR_Y}) in")
    print(f"  Motor R             : ({MOTOR_RX:.1f}, {MOTOR_Y}) in")
    print()
    print(f"  ★ Proximal link L1  : {L1:.3f} in  ({L1/12:.3f} ft)")
    print(f"  ★ Distal   link L2  : {L2:.3f} in  ({L2/12:.3f} ft)")
    print(f"  ★ Total arm reach   : {L1+L2:.3f} in  ({(L1+L2)/12:.3f} ft)")
    print(f"  ★ Target coverage   : {cov:.2%}")
    print()

    n_total = len(vp) + len(sp) if vp is not None else 0
    if n_total > 0:
        sing_pct = len(sp) / n_total
        print(f"  Valid configurations   : {len(vp):,}")
        print(f"  Singular configurations: {len(sp):,}  ({sing_pct:.1%} of reachable)")
        reasons = {}
        for _, _, r in slog:
            key = r.split(":")[0]
            reasons[key] = reasons.get(key, 0) + 1
        for k, v in reasons.items():
            print(f"    └─ {k}: {v:,} configs")

    if vp is not None and len(vp):
        in_tgt = vp[
            (vp[:, 0] >= 0)         & (vp[:, 0] <= FIELD_W) &
            (vp[:, 1] >= TARGET_Y0) & (vp[:, 1] <= TARGET_Y1)
        ]
        print(f"\n  In-target valid pts : {len(in_tgt):,}")
        if len(in_tgt):
            print(f"  x range : {in_tgt[:,0].min():.2f} – {in_tgt[:,0].max():.2f} in")
            print(f"  y range : {in_tgt[:,1].min():.2f} – {in_tgt[:,1].max():.2f} in")
    print("═" * 62)

def visualise(best, results, save_path="linkage_analysis.png"):
    L1_arr  = np.array([r[0] for r in results])
    L2_arr  = np.array([r[1] for r in results])
    cov_arr = np.array([r[2] for r in results])

    L1_u = np.unique(L1_arr)
    L2_u = np.unique(L2_arr)
    n    = len(L1_u)
    heatmap = cov_arr.reshape(n, n)

    vp = best["valid_pts"]
    sp = best["singular_pts"]

    fig = plt.figure(figsize=(18, 7), facecolor="#0d0d14")
    fig.suptitle(
        f"5-Bar Linkage  |  L1={best['L1']:.2f}\"  L2={best['L2']:.2f}\"  "
        f"Target coverage={best['coverage']:.1%}",
        color="white", fontsize=13, fontweight="bold",
    )

    ax1 = fig.add_subplot(1, 3, 1)
    ax2 = fig.add_subplot(1, 3, 2)
    ax3 = fig.add_subplot(1, 3, 3)

    dark = "#0d0d14"
    for ax in (ax1, ax2, ax3):
        ax.set_facecolor(dark)
        for sp_ in ax.spines.values():
            sp_.set_color("#333")
        ax.tick_params(colors="#666")

    if len(vp):
        out = vp[(vp[:, 1] < TARGET_Y0) | (vp[:, 1] > TARGET_Y1)]
        ins = vp[(vp[:, 1] >= TARGET_Y0) & (vp[:, 1] <= TARGET_Y1)]
        if len(out): ax1.scatter(out[:, 0], out[:, 1], s=0.3, c="#003566", alpha=0.25)
        if len(ins): ax1.scatter(ins[:, 0], ins[:, 1], s=0.3, c="#00e5ff", alpha=0.35)
    if len(sp):
        ax1.scatter(sp[:, 0], sp[:, 1], s=2.0, c="#ff4040", alpha=0.6,
                    label="Singular locus", zorder=4)

    ax1.add_patch(patches.Rectangle(                       # full field outline
        (0, 0), FIELD_W, FIELD_H, lw=1.0, ec="#444", fc="none", ls=":"))
    ax1.add_patch(patches.Rectangle(                       # target zone
        (0, TARGET_Y0), TARGET_W, TARGET_H,
        lw=1.5, ec="#ff6b35", fc="#ff6b3311", ls="--"))
    ax1.text(1, TARGET_Y0 + 1, "target zone", color="#ff6b35", fontsize=7)

    ax1.plot(*MOTOR_L, "o", color="#ffe14d", ms=8, zorder=6)
    ax1.plot(*MOTOR_R, "o", color="#ffe14d", ms=8, zorder=6)
    ax1.annotate("M_L", MOTOR_L, xytext=(4, 4),
                 textcoords="offset points", color="#ffe14d", fontsize=7)
    ax1.annotate("M_R", MOTOR_R, xytext=(4, 4),
                 textcoords="offset points", color="#ffe14d", fontsize=7)

    if len(sp):
        ax1.legend(fontsize=7, facecolor="#1a1a24", labelcolor="white",
                   framealpha=0.8, loc="upper right")

    ax1.set_xlim(-3, FIELD_W + 3)
    ax1.set_ylim(-3, FIELD_H + 3)
    ax1.set_aspect("equal")
    ax1.set_title("Full Workspace\ncyan = valid   red = singular", color="white", fontsize=9)
    ax1.set_xlabel("x (in)", color="#aaa", fontsize=8)
    ax1.set_ylabel("y (in)", color="#aaa", fontsize=8)
    cmap = LinearSegmentedColormap.from_list(
        "heat", ["#0d0d14", "#003566", "#00b4d8", "#00e5ff", "#ffffff"])
    im = ax2.imshow(
        heatmap.T, origin="lower",
        extent=[L1_u[0], L1_u[-1], L2_u[0], L2_u[-1]],
        aspect="auto", cmap=cmap, vmin=0, vmax=max(heatmap.max(), 0.01),
    )
    cb = fig.colorbar(im, ax=ax2, fraction=0.046, pad=0.04)
    cb.set_label("Target coverage", color="#aaa", fontsize=8)
    cb.ax.tick_params(colors="#666")
    ax2.plot(best["L1"], best["L2"], "x", color="#ff6b35", ms=11, mew=2.5, zorder=5)
    ax2.set_title("Coverage Heat Map\n(L1 × L2 search space)", color="white", fontsize=9)
    ax2.set_xlabel("L1 — proximal (in)", color="#aaa", fontsize=8)
    ax2.set_ylabel("L2 — distal (in)",   color="#aaa", fontsize=8)
    res = GRID_RES
    nx  = int(np.ceil(TARGET_W / res))
    ny  = int(np.ceil(TARGET_H / res))
    density = np.zeros((nx, ny))

    if len(vp):
        ins = vp[
            (vp[:, 0] >= 0)         & (vp[:, 0] <= FIELD_W) &
            (vp[:, 1] >= TARGET_Y0) & (vp[:, 1] <= TARGET_Y1)
        ]
        for x, y in ins:
            ix = int(x / res)
            iy = int((y - TARGET_Y0) / res)
            if 0 <= ix < nx and 0 <= iy < ny:
                density[ix, iy] += 1
    density = np.log1p(density)

    ax3.imshow(density.T, origin="lower", cmap="inferno",
               extent=[0, TARGET_W, TARGET_Y0, TARGET_Y1], aspect="equal")
    ax3.add_patch(patches.Rectangle(
        (0, TARGET_Y0), TARGET_W, TARGET_H,
        lw=1.5, ec="#ff6b35", fc="none", ls="--"))
    ax3.set_title("Valid-Point Density on Target Zone\n(log scale)",
                  color="white", fontsize=9)
    ax3.set_xlabel("x (in)", color="#aaa", fontsize=8)
    ax3.set_ylabel("y (in)", color="#aaa", fontsize=8)

    plt.tight_layout()
    plt.savefig(save_path, dpi=150, bbox_inches="tight", facecolor=dark)
    print(f"\n  Plot saved → {save_path}")
    plt.show()


if __name__ == "__main__":
    print(f"  Full field  : {FIELD_W:.0f} × {FIELD_H:.0f} in  (2 × 4 ft)")
    print(f"  Target zone : y = {TARGET_Y0:.0f} … {TARGET_Y1:.0f} in  (front 2×2 ft)")
    print(f"  Motor L     : ({MOTOR_LX}, {MOTOR_Y}) in")
    print(f"  Motor R     : ({MOTOR_RX}, {MOTOR_Y}) in\n")
    print("── Coarse search  (L1, L2 ∈ [4, 26] in) ──")
    best_c, results_c = search_linkage_lengths(
        L1_range=(4, 26),
        L2_range=(4, 26),
        n_samples=15,
        theta_steps=100,
        verbose=True,
    )
    print(f"\n  Coarse best → L1={best_c['L1']:.2f}\"  "
          f"L2={best_c['L2']:.2f}\"  coverage={best_c['coverage']:.1%}")
    best_f, _ = fine_tune(best_c, margin=2.0, n_samples=20, theta_steps=150)
    print_report(best_f)
    visualise(best_f, results_c, save_path="linkage_analysis.png")
```
